# ☕ Warm-Up Solution: The Coffee Shop Regulars

1. Check today's visitors against the frequent customer list.
2. Count how many times each frequent customer visited last week, and save the counts to a report file.

## Step 0: Create the data files

```python
# List of the café's frequent customers (one name per line)
with open("frequent_customers.txt", "w") as file:
    file.write("Ana\n")
    file.write("Marcus\n")
    file.write("Priya\n")
    file.write("Diego\n")
    file.write("Lena\n")

# Visit log for last week: day,customer name
with open("visits_last_week.txt", "w") as file:
    file.write("Monday,Ana\n")
    file.write("Monday,Tom\n")
    file.write("Monday,Priya\n")
    file.write("Tuesday,Marcus\n")
    file.write("Tuesday,Ana\n")
    file.write("Wednesday,Sofia\n")
    file.write("Wednesday,Ana\n")
    file.write("Wednesday,Diego\n")
    file.write("Thursday,Priya\n")
    file.write("Thursday,Marcus\n")
    file.write("Friday,Ana\n")
    file.write("Friday,Tom\n")
    file.write("Saturday,Priya\n")
    file.write("Saturday,Kevin\n")
    file.write("Sunday,Marcus\n")

print("Files created!")
```

## Part 1: Which of today's visitors are frequent customers?

```python
todays_visitors = ["Tom", "Priya", "Kevin", "Lena", "Sofia"]

# Read the frequent customer names into a list
frequent = []
with open("frequent_customers.txt", "r") as file:
    for line in file:
        frequent.append(line.strip())   # strip() removes the \n

# Check each visitor against the list
for name in todays_visitors:
    if name in frequent:
        print(f"{name} is a frequent customer! ⭐")
    else:
        print(f"{name} is not on the frequent customer list.")
```

## Part 2: How many times did each frequent customer visit last week?

```python
# Start every frequent customer at 0 visits
visit_counts = {}
for name in frequent:
    visit_counts[name] = 0

# Count visits, but only for frequent customers
with open("visits_last_week.txt", "r") as file:
    for line in file:
        day, customer = line.strip().split(",")
        if customer in visit_counts:
            visit_counts[customer] += 1

# Write the report
with open("visit_report.txt", "w") as report:
    report.write("Frequent Customer Visits - Last Week\n")
    report.write("-" * 36 + "\n")
    for name, count in visit_counts.items():
        report.write(f"{name}: {count}\n")
    report.write("-" * 36 + "\n")
    report.write(f"Total visits: {sum(visit_counts.values())}\n")

# Read the report back and print it
with open("visit_report.txt", "r") as report:
    print(report.read())
```
